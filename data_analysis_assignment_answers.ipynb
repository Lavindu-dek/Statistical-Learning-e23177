### Sample Answer

1. Derivation of the Bias in the MLE Covariance Matrix

We begin with the Maximum Likelihood Estimator for the population covariance matrix $\boldsymbol{\Sigma}$:


$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$


where $\widehat{\boldsymbol{\mu}}_n = \frac{1}{n}\sum_{i=1}^n \mathbf{X}_i$ is the empirical sample mean.

To determine if this estimator is biased, we expand the deviations around the true, hidden population mean $\boldsymbol{\mu}$:


$$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Substituting this back into the sum inside the MLE formula gives:


$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]^T$$

Expanding the product matrix term-by-term yields:


$$= \sum_{i=1}^n (\mathbf{x}_i - \boldsymbol{\mu})(\mathbf{x}_i - \boldsymbol{\mu})^T - \sum_{i=1}^n (\mathbf{x}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{x}_i - \boldsymbol{\mu})^T + \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Notice that $\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$. Using this simplification, the two middle cross-product terms contract:


$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T + n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Now, we apply the expectation operator $\mathbb{E}[\cdot]$ to both sides:


$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^n \mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] - \mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T]$$

By definition of the structural population moments:

* $\mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] = \boldsymbol{\Sigma}$
* $\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T] = \text{Var}(\widehat{\boldsymbol{\mu}}_n) = \frac{1}{n}\boldsymbol{\Sigma}$

Substituting these back into our expectation equation gives:


$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} (n\boldsymbol{\Sigma}) - \frac{1}{n}\boldsymbol{\Sigma} = \boldsymbol{\Sigma} - \frac{1}{n}\boldsymbol{\Sigma} = \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma}$$

Because $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] \neq \boldsymbol{\Sigma}$, the estimator is systematically downward biased. This occurs because calculating variations against the *sample* mean rather than the *true* population mean artificially underrepresents the real spatial scattering of the system.

**Bessel's Correction Elimination of Bias**

To yield an unbiased estimator ($\mathbf{S}$), we scale the maximum likelihood matrix by the reciprocal of the bias factor ($\frac{n}{n-1}$):


$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right) = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Evaluating the expectation confirms its lack of bias:


$$\mathbb{E}[\mathbf{S}] = \frac{n}{n-1} \mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{n}{n-1} \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma} = \boldsymbol{\Sigma}$$

---

**Operational Error Diagnostics**

* **Type I Error ($\alpha$):** A **False Alarm**. This happens when the diagnostic system concludes that the structure is damaged or unstable (rejects $H_0$), when in reality the physical system remains completely nominal and healthy.
* **Type II Error ($\beta$):** A **Missed Detection**. This occurs when a structural shift or localized crack has compromised the asset (rejects $H_1$), but the diagnostic system fails to alert the team (fails to reject $H_0$).

**Consequence of an Ultra-Conservative Threshold ($\alpha = 0.0001$)**

When $\alpha$ is reduced dramatically to eliminate false alarms, the critical value boundary of the test statistic is pushed deep into the extreme tails of the baseline distribution.

1. **Impact on Statistical Power ($1-\beta$):** Because the threshold for detecting anomalies becomes incredibly stringent, minor or emerging structural shifts will fail to cross the new partition line. Therefore, $\beta$ (missed detections) will sharply escalate, causing the **statistical power ($1-\beta$) of the model to collapse**.
2. **Impact on Confidence Ellipsoids:** The confidence level of the multi-dimensional boundary scales as $1-\alpha = 0.9999$. This means the **geometric volume of the healthy operational ellipsoid expands significantly**. The boundary swells to encompass almost all variation, making the system highly insensitive to genuine baseline degradation.

## **Part B: Theoretical Extension (Asymptotic Distributions & Slutsky's Theorem)**

The Lindeberg-Lévy Multivariate Central Limit Theorem establishes that as the sample snapshot size approaches infinity ($n \to \infty$), the standardized difference between the empirical mean and population mean converges in distribution to a Multivariate Gaussian:

$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$



However, the true population covariance matrix $\boldsymbol{\Sigma}$ is a latent, unobservable parameter.
* State **Slutsky’s Theorem** rigorously.
* Prove how Slutsky's Theorem justifies substituting the empirical sample covariance matrix $\mathbf{S}$ for the latent population matrix $\boldsymbol{\Sigma}$ as $n \to \infty$, thereby operationalizing the practical parametric modeling distribution used in your dashboard:

$$\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left(\widehat{\boldsymbol{\mu}}_n, \frac{1}{n}\mathbf{S}\right)$$







### Sample Answer

Let $\{X_n\}$ and $\{Y_n\}$ be sequences of random variables (or vectors). If $X_n$ converges in distribution to a random variable $X$ ($X_n \xrightarrow{d} X$), and $Y_n$ converges in probability to a constant scalar or matrix $c$ ($Y_n \xrightarrow{p} c$), then:

* $X_n + Y_n \xrightarrow{d} X + c$
* $X_n Y_n \xrightarrow{d} cX$
* $X_n / Y_n \xrightarrow{d} X/c$ (provided $c \neq 0$)

**Operational Distribution Justification**

The Lindeberg-Lévy Multivariate CLT asserts that:


$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

Multiplying by the inverse matrix square root of the population covariance normalizes the space:


$$\boldsymbol{\Sigma}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \mathscr{N}(\mathbf{0}, \mathbf{I})$$

By the Weak Law of Large Numbers (WLLN), the unbiased empirical sample covariance matrix $\mathbf{S}$ is a continuous, consistent estimator of the population matrix $\boldsymbol{\Sigma}$. Therefore, as $n \to \infty$, $\mathbf{S}$ converges in probability to the true constant profile:


$$\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma} \implies \mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$$

Applying **Slutsky's Theorem**, we multiply the sequence converging in distribution by the sequence converging in probability to a constant matrix:


$$\mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2} \cdot \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma}) = \mathscr{N}(\mathbf{0}, \mathbf{I})$$

Because the target boundary converges asymptotically to a standard Gaussian space, we scale the localized coordinate parameters back to the empirical matrix metric:


$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \sim \mathscr{N}(\mathbf{0}, \mathbf{S}) \implies \widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left(\boldsymbol{\mu}, \frac{1}{n}\mathbf{S}\right)$$

This mathematically validates substituting $\mathbf{S}$ for $\boldsymbol{\Sigma}$ without corrupting the stability of the asymptotic diagnostic envelope.

***Scenario:** You are provided an industrial dataset containing $n = 5,000$ snapshots across $m = 3$ sensor channels measuring physical strain. Implement a Multivariate Analysis of Variance (MANOVA) tracking framework from scratch (or leveraging `statsmodels`) to calculate **Wilks' Lambda ($\Lambda$)**.

$$\Lambda = \frac{|\mathbf{W}|}{|\mathbf{B} + \mathbf{W}|}$$

$$\chi^2_{\text{calc}} = -\left(n - 1 - \frac{m + g}{2}\right) \ln \Lambda$$

Report your exact calculated Wilks' $\Lambda$, the Bartlett $\chi^2$ test statistic, and its corresponding $p$-value under $\chi^2 \sim \chi^2_{m(g-1)}$. ### **Sample Answer**

````python
import numpy as np
import pandas as pd
import scipy.stats as stats

# ─── Synthetic Data Generator (DO NOT MODIFY) ─────────────────────────────────
np.random.seed(42)
n_samples  = 5000
n_features = 3

base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)
# Subtle drift injected into the last 1,000 rows
base_data[4000:, 0] += 0.015   # Sensor 1 drift
base_data[4000:, 2] -= 0.010   # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ──────────────────────────────────────────────────────────────────────────────


def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Tests whether the joint mean vector is stationary across g consecutive
    non-overlapping blocks using Wilks' Lambda and Bartlett's Chi-Square
    approximation.

    Parameters
    ----------
    df       : DataFrame with numeric sensor columns.
    g_chunks : Number of equal-size consecutive partitions.

    Returns
    -------
    dict with keys: wilks_lambda, chi2_stat, degrees_of_freedom, p_value,
                    reject_H0, chunk_means.
    """
    X = df.values                      # (n, m) array
    n, m = X.shape

    if n % g_chunks != 0:
        raise ValueError(f"n={n} is not divisible by g={g_chunks}.")

    chunk_size = n // g_chunks

    # ── Step 1: Partition data into g consecutive blocks ──────────────────────
    chunks = [X[j * chunk_size:(j + 1) * chunk_size] for j in range(g_chunks)]

    # ── Step 2: Global and per-block mean vectors ─────────────────────────────
    global_mean  = np.mean(X, axis=0)                     # shape (m,)
    chunk_means  = [np.mean(c, axis=0) for c in chunks]   # list of (m,) arrays

    # ── Step 3: Within-block scatter matrix W ─────────────────────────────────
    # W = sum over blocks of sum of squared deviations from the block mean
    W = np.zeros((m, m))
    for c, mu_j in zip(chunks, chunk_means):
        deviations = c - mu_j          # (chunk_size, m)
        W += deviations.T @ deviations # accumulate outer products

    # ── Step 4: Between-block scatter matrix B ────────────────────────────────
    # B = sum over blocks of n_j * (mu_j - mu_global)(mu_j - mu_global)^T
    B = np.zeros((m, m))
    for mu_j in chunk_means:
        diff = (mu_j - global_mean).reshape(-1, 1)   # column vector
        B += chunk_size * (diff @ diff.T)

    # ── Step 5: Wilks' Lambda ─────────────────────────────────────────────────
    det_W     = np.linalg.det(W)
    det_BplusW = np.linalg.det(B + W)
    wilks_lambda = det_W / det_BplusW

    # ── Step 6: Bartlett's Chi-Square transformation ──────────────────────────
    df_chi2   = m * (g_chunks - 1)                              # degrees of freedom
    bartlett_multiplier = -(n - 1 - (m + g_chunks) / 2.0)
    chi2_stat = bartlett_multiplier * np.log(wilks_lambda)
    p_value   = 1.0 - stats.chi2.cdf(chi2_stat, df=df_chi2)

    alpha = 0.05
    reject_H0 = p_value < alpha

    # ── Report ─────────────────────────────────────────────────────────────────
    print("=" * 60)
    print("  First-Moment Homogeneity Test — MANOVA (Wilks' Lambda)")
    print("=" * 60)
    print(f"  Number of blocks  (g) : {g_chunks}")
    print(f"  Block size            : {chunk_size}")
    print(f"  Feature dimensions (m): {m}")
    print("-" * 60)
    print(f"  Wilks' Lambda  (Λ)    : {wilks_lambda:.6f}")
    print(f"  Bartlett χ² statistic : {chi2_stat:.4f}")
    print(f"  Degrees of freedom    : {df_chi2}")
    print(f"  p-value               : {p_value:.6e}")
    print("-" * 60)
    decision = "REJECT H₀ — Mean shift detected" if reject_H0 else "FAIL TO REJECT H₀ — Means appear stationary"
    print(f"  Decision (α=0.05)     : {decision}")
    print("=" * 60)

    return {
        "wilks_lambda"      : wilks_lambda,
        "chi2_stat"         : chi2_stat,
        "degrees_of_freedom": df_chi2,
        "p_value"           : p_value,
        "reject_H0"         : reject_H0,
        "chunk_means"       : chunk_means,
        "W_matrix"          : W,
        "B_matrix"          : B,
    }


# ── Execute ────────────────────────────────────────────────────────────────────
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)
````

### Interpretation of Results

Because the synthetic data intentionally injects a small drift into the last 1,000 observations, the expected outcome is:

- **Wilks' Lambda** noticeably below 1.0 (a value of 1.0 would indicate zero between-group separation).
- **Bartlett $\chi^2$** substantially exceeds the critical value $\chi^2_{m(g-1),\,0.95} = \chi^2_{12,\,0.95} \approx 21.03$.
- **p-value** falls well below $\alpha = 0.05$.

**Diagnostic Conclusion:** We reject $H_0$. The joint mean vector is **not** stationary across the five temporal blocks. A significant baseline shift has been detected, violating the first-moment stationarity assumption required before applying PCA or Factor Analysis globally.

## **Sample Answer**


### 1)
**Setup.** The spectral decomposition of $\boldsymbol{\Sigma}$ yields $\boldsymbol{\Sigma} = \mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T$ where columns of $\mathbf{P}$ are orthonormal eigenvectors and $\boldsymbol{\Lambda} = \text{diag}(\lambda_1, \ldots, \lambda_m)$. The projected scores are $\mathbf{Z}_i = \mathbf{P}^T\widetilde{\mathbf{X}}_i$.

**Derivation:**

$$\mathbb{E}[\mathbf{Z}_i\mathbf{Z}_i^T] = \mathbb{E}\bigl[\mathbf{P}^T\widetilde{\mathbf{X}}_i\widetilde{\mathbf{X}}_i^T\mathbf{P}\bigr] = \mathbf{P}^T\,\mathbb{E}[\widetilde{\mathbf{X}}_i\widetilde{\mathbf{X}}_i^T]\,\mathbf{P} = \mathbf{P}^T\boldsymbol{\Sigma}\mathbf{P}$$

Substituting the spectral decomposition:

$$= \mathbf{P}^T(\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T)\mathbf{P} = (\mathbf{P}^T\mathbf{P})\,\boldsymbol{\Lambda}\,(\mathbf{P}^T\mathbf{P}) = \mathbf{I}\,\boldsymbol{\Lambda}\,\mathbf{I} = \boldsymbol{\Lambda}$$

using the orthonormality condition $\mathbf{P}^T\mathbf{P} = \mathbf{I}$.

**Geometric interpretation.** The off-diagonal entries of $\boldsymbol{\Lambda}$ are all zero, which means $\text{Cov}(Z_{i,j}, Z_{i,k}) = 0$ for $j \neq k$. The PCA rotation completely **de-correlates** the original sensor channels. Each principal component score is linearly independent of every other, and the information content of each axis is summarized solely by its eigenvalue $\lambda_j$.

---

### 2)

**Proof that total variance is preserved under the rotation $\mathbf{P}$:**

Using the cyclic property of the matrix trace ($\text{tr}(\mathbf{ABC}) = \text{tr}(\mathbf{CAB})$):

$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T) = \text{tr}(\boldsymbol{\Lambda}\underbrace{\mathbf{P}^T\mathbf{P}}_{=\mathbf{I}}) = \text{tr}(\boldsymbol{\Lambda}) = \sum_{j=1}^m \lambda_j$$

The total system energy is **unchanged** by the rotation; it has simply been redistributed across the new orthogonal axes.

**Variance ratios for a truncated model retaining $k$ components:**

$$\Phi(k) = \frac{\sum_{j=1}^k \lambda_j}{\sum_{j=1}^m \lambda_j} \quad\text{(Cumulative Explained Variance Ratio)}$$

$$\Psi(k) = 1 - \Phi(k) = \frac{\sum_{j=k+1}^m \lambda_j}{\sum_{j=1}^m \lambda_j} \quad\text{(Residual Unexplained Variance Ratio)}$$

---

### 3)

**Reconstruction and residual vectors:**

$$\widehat{\mathbf{x}}_i = \widehat{\boldsymbol{\mu}}_n + \widehat{\mathbf{P}}_k\mathbf{z}_{i,k} \quad (\text{projection back into sensor space})$$

$$\mathbf{e}_i = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) - \widehat{\mathbf{P}}_k\mathbf{z}_{i,k} = \widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}$$

**Pythagorean identity proof.** Because $\widehat{\mathbf{P}}_k$ and $\widehat{\mathbf{P}}_{m-k}$ span **orthogonal** subspaces ($\widehat{\mathbf{P}}_k^T\widehat{\mathbf{P}}_{m-k} = \mathbf{0}$), the cross-product term vanishes:

$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \|\widehat{\mathbf{P}}_k\mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}\|^2 = \|\mathbf{z}_{i,k}\|^2 + \underbrace{2\,\mathbf{z}_{i,k}^T\widehat{\mathbf{P}}_k^T\widehat{\mathbf{P}}_{m-k}\mathbf{z}_{i,m-k}}_{= 0} + \|\mathbf{z}_{i,m-k}\|^2$$

$$\boxed{\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2}$$

**Diagnostic statistic comparison:**

| Statistic | Formula | Physical Meaning | Best Suited For |
|---|---|---|---|
| **Hotelling's $T^2$** | $T^2_i = \sum_{j=1}^k z_{i,j}^2 / \widehat{\lambda}_j$ | Standardized distance *within* the principal subspace | Shifts along known structural response modes (e.g., extreme wind loads driving expected vibration patterns) |
| **$Q$ Statistic (SPE)** | $Q_i = \sum_{j=k+1}^m z_{i,j}^2$ | Residual energy *outside* the learned subspace | Novel, unexpected fault directions (e.g., a fatigue crack opening a new load path not seen in training data) |

**Matching to scenarios:**
- **Extreme wind load:** The load activates the dominant structural modes already captured in the principal subspace. $T^2$ is more sensitive because the anomaly moves along the known response directions.
- **Internal fatigue crack:** Damage changes the load transfer path, generating sensor responses that lie *outside* the learned subspace. $Q$ (SPE) is more sensitive because it monitors residual, unexplained energy.

````python
import numpy as np
import pandas as pd


class PCADiagnosticsEngine:
    """
    Decomposes the unbiased sample covariance matrix and computes
    Hotelling's T² and the Q (SPE) statistic for k in {1, 2, ..., m-1}.
    """

    def __init__(self, X: np.ndarray):
        """
        Parameters
        ----------
        X : array of shape (n, m) — raw sensor snapshots.
        """
        self.X = X
        self.n, self.m = X.shape

    def fit_and_evaluate(self) -> dict:
        """
        1. Centers the data.
        2. Computes the unbiased sample covariance S.
        3. Spectral decomposition — eigenvalues sorted in descending order.
        4. For each k in {1 … m-1}: computes T² and Q arrays + their means.
        Returns a dictionary of all computed metrics.
        """
        n, m = self.n, self.m

        # ── 1. Center ──────────────────────────────────────────────────────────
        mu_hat    = np.mean(self.X, axis=0)
        X_centered = self.X - mu_hat

        # ── 2. Unbiased sample covariance (ddof=1) ────────────────────────────
        S = np.cov(self.X, rowvar=False, ddof=1)

        # ── 3. Spectral decomposition (eigh is numerically stable for Hermitian)
        raw_eigenvalues, raw_eigenvectors = np.linalg.eigh(S)
        sort_idx    = np.argsort(raw_eigenvalues)[::-1]          # descending
        lambda_hat  = np.clip(raw_eigenvalues[sort_idx], 1e-15, None)
        P_hat       = raw_eigenvectors[:, sort_idx]

        # ── 4. Project all samples onto the full principal space ──────────────
        Z_full = X_centered @ P_hat    # (n, m)

        # ── 5. Loop over truncation cutoffs ───────────────────────────────────
        k_range      = np.arange(1, m)   # {1, 2, ..., m-1}
        mean_T2_list = []
        mean_Q_list  = []

        print(f"{'k':>4} | {'Mean T²':>12} | {'Mean Q (SPE)':>14}")
        print("-" * 36)

        for k in k_range:
            # Hotelling's T²: standardized distance inside the k-subspace
            Z_k  = Z_full[:, :k]                      # retained scores
            T2_i = np.sum((Z_k ** 2) / lambda_hat[:k], axis=1)

            # Q statistic: residual energy in the discarded subspace
            Z_res = Z_full[:, k:]                     # residual scores
            Q_i   = np.sum(Z_res ** 2, axis=1)

            mean_T2_list.append(np.mean(T2_i))
            mean_Q_list.append(np.mean(Q_i))

            print(f"  {k:>2} | {np.mean(T2_i):>12.4f} | {np.mean(Q_i):>14.4f}")

        return {
            "mean_vector"       : mu_hat,
            "covariance_S"      : S,
            "eigenvalues"       : lambda_hat,
            "eigenvectors_P"    : P_hat,
            "scores_Z"          : Z_full,
            "k_range"           : k_range,
            "mean_T2_per_k"     : np.array(mean_T2_list),
            "mean_Q_per_k"      : np.array(mean_Q_list),
        }


# ── Synthetic 4-sensor baseline archive ────────────────────────────────────────
np.random.seed(0)
n_baseline = 3000
Sigma_true  = np.array([[1.0, 0.7, 0.3, 0.1],
                         [0.7, 1.0, 0.5, 0.2],
                         [0.3, 0.5, 1.0, 0.4],
                         [0.1, 0.2, 0.4, 1.0]])
X_baseline  = np.random.multivariate_normal(np.zeros(4), Sigma_true, size=n_baseline)

pca_engine = PCADiagnosticsEngine(X_baseline)
pca_results = pca_engine.fit_and_evaluate()
````

# **Latent Subspace Decomposition via Factor Analysis (FA)**

The objective is to master the generative model equations of Factor Analysis, derive factor scores from scratch using conditional expectations (Thomson's method), and numerically isolate sensor malfunctions from true structural damage by tracking communalities ($h^2$) and uniquenesses ($\varphi^2$).

## **Part A: Theoretical Fundamentals (Generative Model & Commonalities)**

1. Consider your structured data setup where $\mathbf{Z}_i \in \mathbb{R}^m$ is a normalized, zero-mean baseline feature vector whose population covariance matrix represents the universal correlation matrix $\mathbf{R} = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T]$. The Factor Analysis model posits the linear generative structure:
$$\mathbf{Z}_i = \boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i$$
where $\mathbf{F}_i \in \mathbb{R}^k$ are the orthogonal common factors ($\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] = \mathbf{I}_{k \times k}$) and $\boldsymbol{\epsilon}_i \in \mathbb{R}^m$ is the diagonalized measurement noise matrix ($\mathbb{E}[\boldsymbol{\epsilon}_i\boldsymbol{\epsilon}_i^T] = \boldsymbol{\Psi} = \text{diag}(\varphi_1^2, \dots, \varphi_m^2)$).
* Utilizing the constraint that the latent factors and errors are independent ($\mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] = \mathbf{0}$), prove the **Fundamental Equation of Factor Analysis**:

$$\mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$


* Focus on the $j$-th diagonal entry of this decomposition. Explicitly define the **Communality ($h_j^2$)** and **Uniqueness ($\varphi_j^2$)** parameters, and explain their physical interpretations in an asset diagnostics system.


2. In Factor Analysis, we often apply an orthogonal coordinate rotation to the factor loading matrix (such as the **Varimax** rotation criterion) to improve model interpretability.
* Contrast how a raw principal component loading vector differs conceptually from a Varimax-rotated factor loading vector.
* Explain why orthogonal rotations can alter the individual columns of the loading matrix $\boldsymbol{\Lambda}$ without changing the calculated communality ($h_j^2$) of any sensor or modifying the global covariance approximation $\mathbf{R} \approx \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$.



3. Because the common factors $\mathbf{F}_i$ are unobservable latent parameters, they cannot be directly extracted via geometric projections like PCA. Instead, we must treat them as hidden random vectors and calculate their realization using **Thomson’s Regression Method**, which relies on the conditional expectation $\mathbf{f}_i = \mathbb{E}[\mathbf{F}_i | \mathbf{z}_i]$.
* Construct the joint partitioned vector $\mathbf{Y}_i = [\mathbf{Z}_i^T, \mathbf{F}_i^T]^T$ and prove that its joint population covariance matrix expands exactly to:

$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbf{R} & \boldsymbol{\Lambda} \\ \boldsymbol{\Lambda}^T & \mathbf{I}_{k \times k} \end{bmatrix}$$


* Under the standard assumption of a Multivariate Normal distribution, the conditional mean of a partitioned system is governed by the projection identity $\mathbb{E}[\mathbf{X}_2 | \mathbf{x}_1] = \boldsymbol{\mu}_2 + \boldsymbol{\Sigma}_{21} \boldsymbol{\Sigma}_{11}^{-1} (\mathbf{x}_1 - \boldsymbol{\mu}_1)$. Using this theorem, derive the final score matrix equation for Thomson’s estimator:

$$\mathbf{f}_i = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i$$


## **Sample Answer**

### 1)

**Generative model:** $\mathbf{Z}_i = \boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i$, with:
- $\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] = \mathbf{I}_k$ (orthonormal factors)
- $\mathbb{E}[\boldsymbol{\epsilon}_i\boldsymbol{\epsilon}_i^T] = \boldsymbol{\Psi} = \text{diag}(\varphi_1^2, \ldots, \varphi_m^2)$ (diagonal noise)
- $\mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T] = \mathbf{0}$ (noise independent of factors)

**Proof:**

$$\mathbf{R} = \mathbb{E}[\mathbf{Z}_i\mathbf{Z}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)^T]$$

Expanding:

$$= \boldsymbol{\Lambda}\,\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T]\,\boldsymbol{\Lambda}^T + \boldsymbol{\Lambda}\underbrace{\mathbb{E}[\mathbf{F}_i\boldsymbol{\epsilon}_i^T]}_{=\mathbf{0}} + \underbrace{\mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T]}_{=\mathbf{0}}\boldsymbol{\Lambda}^T + \mathbb{E}[\boldsymbol{\epsilon}_i\boldsymbol{\epsilon}_i^T]$$

$$= \boldsymbol{\Lambda}\mathbf{I}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi} = \boxed{\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}}$$

**Communality and Uniqueness on the $j$-th channel:**

The $j$-th diagonal entry of $\mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$ gives:

$$R_{jj} = \underbrace{\sum_{r=1}^k \lambda_{jr}^2}_{h_j^2 \;\text{communality}} + \underbrace{\varphi_j^2}_{\text{uniqueness}} = 1$$

- **Communality $h_j^2$** — the fraction of `Sensor_j`'s total variance that is **shared with** (explained by) the common latent factors. A high value means the sensor tracks the underlying physical processes faithfully.
- **Uniqueness $\varphi_j^2$** — the fraction remaining after the shared part is removed. This is pure localized noise: electrical interference, sensor drift, or instrument faults that are **not** visible in any other channel.

---

### 2)

**PCA eigenvectors** are computed to greedily maximize captured variance, so the first component grabs a mix of all sensors simultaneously. The loadings often show many moderate values spread across every sensor — a "smeared" representation that is hard to interpret physically.

**Varimax rotation** post-multiplies the loading matrix by an orthogonal rotation matrix $\mathbf{T}$ ($\boldsymbol{\Lambda}_{\text{rot}} = \boldsymbol{\Lambda}\mathbf{T}$) to maximise the within-column variance of squared loadings. The result is **simple structure**: each factor shows high loadings on a small, well-defined subset of sensors and near-zero loadings elsewhere. This makes physical attribution unambiguous.

**Why communalities are rotation-invariant:**

$$h_j^2 = \sum_{r=1}^k \lambda_{jr}^2 = \|\boldsymbol{\lambda}_j\|^2 = \|\boldsymbol{\lambda}_j\mathbf{T}\|^2$$

because $\mathbf{T}^T\mathbf{T} = \mathbf{I}$, so an orthogonal transformation preserves Euclidean norms. Similarly, the **total covariance reconstruction** $\boldsymbol{\Lambda}_{\text{rot}}\boldsymbol{\Lambda}_{\text{rot}}^T = \boldsymbol{\Lambda}\mathbf{T}\mathbf{T}^T\boldsymbol{\Lambda}^T = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$ is unchanged.

---

### 3)

**Joint covariance of $\mathbf{Y}_i = [\mathbf{Z}_i^T, \mathbf{F}_i^T]^T$:**

$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbb{E}[\mathbf{Z}_i\mathbf{Z}_i^T] & \mathbb{E}[\mathbf{Z}_i\mathbf{F}_i^T] \\ \mathbb{E}[\mathbf{F}_i\mathbf{Z}_i^T] & \mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] \end{bmatrix}$$

Computing the off-diagonal block:

$$\mathbb{E}[\mathbf{Z}_i\mathbf{F}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)\mathbf{F}_i^T] = \boldsymbol{\Lambda}\underbrace{\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T]}_{=\mathbf{I}} + \underbrace{\mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T]}_{=\mathbf{0}} = \boldsymbol{\Lambda}$$

Therefore:

$$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbf{R} & \boldsymbol{\Lambda} \\ \boldsymbol{\Lambda}^T & \mathbf{I}_{k\times k} \end{bmatrix}$$

**Thomson's estimator via the conditional mean formula:**

The conditional expectation of $\mathbf{F}_i$ given the observation $\mathbf{z}_i$ (with both having zero mean) is:

$$\mathbb{E}[\mathbf{F}_i | \mathbf{z}_i] = \boldsymbol{\Sigma}_{21}\boldsymbol{\Sigma}_{11}^{-1}\mathbf{z}_i = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i$$

$$\boxed{\mathbf{f}_i = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i}$$

This is the **minimum mean-squared error** (MMSE) linear estimate of the latent factors given the observed sensor readings.



The class below encapsulates all three pipeline stages:

| Stage | Description |
|---|---|
| **Z-Score Mapping** | Standardise columns; guard zero-variance channels |
| **Dimensionality Constraints** | Validate $n > m$ and $k < m$ |
| **Information Decomposition** | Fit FA (Varimax), extract $h^2$, $\varphi^2$, Thomson scores $F$ |

````python
class FactorAnalysisEngine:
    """
    End-to-end Factor Analysis pipeline.

    Stages
    ------
    1. Z-score standardisation with zero-variance guardrail.
    2. Dimensionality constraint validation.
    3. FA decomposition (SVD / MLE) + Varimax rotation.
    4. Communality h², uniqueness φ², and Thomson score computation.
    5. 2×2 Plotly diagnostic dashboard.
    """

    def __init__(self, df: pd.DataFrame):
        self.df = df

    # ─────────────────────────────────────────────────────────────────────────
    # STAGE 1 · Z-Score Mapping Layer
    # ─────────────────────────────────────────────────────────────────────────
    @staticmethod
    def _zscore_standardise(X: np.ndarray) -> tuple:
        """
        Standardise columns to zero mean / unit variance.
        Zero-variance channels are floored to 1e-15 to prevent divide-by-zero.

        Returns
        -------
        Z       : standardised array  (n, m)
        mu_hat  : column means        (m,)
        std_hat : column std devs     (m,)  (already guarded)
        """
        mu_hat  = X.mean(axis=0)
        std_hat = X.std(axis=0, ddof=1)           # unbiased std

        # ── Zero-variance guardrail ────────────────────────────────────────────
        n_zero = (std_hat == 0).sum()
        if n_zero > 0:
            print(f"[WARNING] {n_zero} zero-variance channel(s) detected. "
                   "Standard deviation floored to 1e-15.")
        std_hat = np.where(std_hat == 0, 1e-15, std_hat)

        Z = (X - mu_hat) / std_hat
        return Z, mu_hat, std_hat

    # ─────────────────────────────────────────────────────────────────────────
    # STAGE 2 · Dimensionality Constraint Layer
    # ─────────────────────────────────────────────────────────────────────────
    @staticmethod
    def _validate_dimensions(n: int, m: int, k: int) -> None:
        """
        Enforce two hard constraints before decomposition.

        Rule 1: n > m  — the snapshot count must exceed the feature count so
                          that the empirical correlation matrix R is full-rank.
        Rule 2: k < m  — the requested latent dimension must be strictly
                          smaller than the observable feature space.
        """
        if n <= m:
            raise ValueError(
                f"Dimensionality Constraint Violation (Rule 1): "
                f"Snapshot count n={n} must be STRICTLY GREATER than "
                f"feature count m={m}. Collect more observations or reduce "
                f"the sensor channel set before proceeding."
            )
        if k >= m:
            raise ValueError(
                f"Dimensionality Constraint Violation (Rule 2): "
                f"Requested latent factor count k={k} must be STRICTLY LESS "
                f"than feature count m={m}. Reduce k to at most {m - 1}."
            )

    # ─────────────────────────────────────────────────────────────────────────
    # STAGE 3 · Information Decomposition + Score Projection
    # ─────────────────────────────────────────────────────────────────────────
    def fit(
        self,
        k: int,
        columns: Optional[Sequence[str]] = None,
        show_plot: bool = True
    ) -> Dict[str, Any]:
        """
        Run the full FA pipeline and render the 2×2 diagnostic dashboard.

        Parameters
        ----------
        k         : number of latent factors to extract.
        columns   : optional list of column names; defaults to all numeric columns.
        show_plot : render the Plotly dashboard (set False for unit-testing).

        Returns
        -------
        dict with keys:
            Z, mu, std, R,
            lambda_rotated, communality_h2, uniqueness_phi2,
            factor_scores_F, factor_variance,
            sensors, k
        """
        # ── Select numeric columns ─────────────────────────────────────────────
        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
        else:
            target_cols = list(columns)

        X    = self.df[target_cols].dropna().values   # (n, m)
        n, m = X.shape

        # ── Stage 2: Validate dimensions ───────────────────────────────────────
        self._validate_dimensions(n, m, k)

        # ── Stage 1: Z-score standardisation ──────────────────────────────────
        Z, mu_hat, std_hat = self._zscore_standardise(X)

        # Empirical correlation matrix (= cov of Z-scored data)
        R_matrix = np.corrcoef(X, rowvar=False)       # (m, m)

        # ── Stage 3a: Fit Factor Analysis (SVD backend) + Varimax rotation ────
        fa_model = FactorAnalysis(
            n_components=k,
            rotation='varimax',
            random_state=42
        )
        fa_model.fit(Z)

        # lambda_rotated  : (m, k)  — rotated loading matrix Λ_rot
        # psi_diagonal    : (m,)    — uniqueness / noise variance per sensor
        lambda_rotated = fa_model.components_.T          # sklearn stores (k, m)
        psi_diagonal   = fa_model.noise_variance_

        # ── Stage 3b: Communality and uniqueness vectors ───────────────────────
        # h²_j  = sum_r λ²_{j,r}  (row-wise sum of squared loadings)
        # φ²_j  = 1 - h²_j        (residual unexplained channel variance)
        communality_h2  = (lambda_rotated ** 2).sum(axis=1)   # (m,)
        uniqueness_phi2 = psi_diagonal                         # (m,)  from model

        # ── Stage 3c: Thomson MMSE regression scores ───────────────────────────
        # F = Z · R⁻¹ · Λ_rot      shape (n, k)
        R_inv = np.linalg.inv(R_matrix)
        F_scores = Z @ R_inv @ lambda_rotated              # (n, k)

        # ── Performance telemetry ──────────────────────────────────────────────
        factor_variance = np.var(F_scores, axis=0, ddof=1)   # unbiased
        avg_communality = communality_h2.mean()  * 100
        avg_uniqueness  = uniqueness_phi2.mean() * 100

        print("\n" + "═" * 62)
        print("  FA PIPELINE — PERFORMANCE DIAGNOSTIC TELEMETRY")
        print("═" * 62)
        print(f"  Sensors (m)            : {m}")
        print(f"  Latent factors (k)     : {k}")
        print(f"  Observations (n)       : {n}")
        print("─" * 62)
        print(f"  Average System Communality  h²  : {avg_communality:>7.2f} %")
        print(f"  Average System Uniqueness   φ²  : {avg_uniqueness:>7.2f} %")
        print("─" * 62)
        print("  Per-sensor breakdown:")
        print(f"  {'Sensor':<12} {'h² (%)':<12} {'φ² (%)':<12}")
        for col, h2, phi2 in zip(target_cols, communality_h2, uniqueness_phi2):
            print(f"  {col:<12} {h2*100:<12.2f} {phi2*100:<12.2f}")
        print("─" * 62)
        print("  Factor score empirical variance (ddof=1):")
        for r, v in enumerate(factor_variance, 1):
            print(f"    Factor {r}  :  {v:.4f}")
        print("═" * 62)

        # ── Dashboard ──────────────────────────────────────────────────────────
        if show_plot:
            self._render_dashboard(
                target_cols, k, lambda_rotated,
                communality_h2, uniqueness_phi2, factor_variance
            )

        return dict(
            Z=Z, mu=mu_hat, std=std_hat, R=R_matrix,
            lambda_rotated=lambda_rotated,
            communality_h2=communality_h2,
            uniqueness_phi2=uniqueness_phi2,
            factor_scores_F=F_scores,
            factor_variance=factor_variance,
            sensors=target_cols, k=k
        )

    # ─────────────────────────────────────────────────────────────────────────
    # STAGE 4 · 2×2 Plotly Diagnostic Dashboard
    # ─────────────────────────────────────────────────────────────────────────
    @staticmethod
    def _render_dashboard(
        sensor_labels: list,
        k: int,
        lambda_rotated: np.ndarray,
        communality_h2: np.ndarray,
        uniqueness_phi2: np.ndarray,
        factor_variance: np.ndarray
    ) -> None:
        """
        Renders the 2×2 diagnostic canvas.

        Layout
        ──────
        (1,1) Structural Loadings Heatmap  |  (1,2) Variance Partitioning Bar
        (2,1) Sensor Uniqueness Line       |  (2,2) Latent Factor Variance Bar
        """
        factor_labels = [f"Factor {r+1}" for r in range(k)]

        # ── Canvas scaffold ────────────────────────────────────────────────────
        fig = make_subplots(
            rows=2, cols=2,
            horizontal_spacing=0.24,
            vertical_spacing=0.28,
            subplot_titles=(
                "Structural Loadings Matrix |Λ_rotated|",
                "Variance Partitioning: Communality vs Uniqueness",
                "Sensor Uniqueness Noise Floor (φ²)",
                "Latent Factor Scores — Empirical Variance"
            )
        )

        # ── (1,1) Loadings Heatmap — left-anchored colorbar ────────────────────
        fig.add_trace(
            go.Heatmap(
                z=np.abs(lambda_rotated),
                x=factor_labels,
                y=sensor_labels,
                colorscale='YlOrRd',
                colorbar=dict(
                    title="Loading",
                    x=-0.15,          # anchored left of the canvas
                    len=0.38,         # height spans row 1
                    y=0.78,
                    yanchor="middle",
                    xanchor="right"
                ),
                showscale=True,
                showlegend=False
            ),
            row=1, col=1
        )
        fig.update_xaxes(title_text="Latent Factors", row=1, col=1)
        fig.update_yaxes(title_text="Sensor Channel",  row=1, col=1)

        # ── (1,2) Stacked Bar: Communality + Uniqueness = 100% ─────────────────
        fig.add_trace(
            go.Bar(
                y=sensor_labels,
                x=communality_h2 * 100,
                name="Communality h² — Shared Structure",
                orientation='h',
                marker=dict(color='#1f77b4'),
                legendgroup="comm"
            ),
            row=1, col=2
        )
        fig.add_trace(
            go.Bar(
                y=sensor_labels,
                x=uniqueness_phi2 * 100,
                name="Uniqueness φ² — Isolated Noise",
                orientation='h',
                marker=dict(color='#ff7f0e'),
                legendgroup="uniq_bar"
            ),
            row=1, col=2
        )
        fig.update_layout(barmode='stack')
        fig.update_xaxes(
            title_text="Variance Allocation (%)",
            range=[0, 100],
            row=1, col=2
        )

        # ── (2,1) Uniqueness Line — dot-dashed red with × markers ───────────────
        fig.add_trace(
            go.Scatter(
                x=sensor_labels,
                y=uniqueness_phi2,
                mode='lines+markers',
                name="Uniqueness φ² Profile",
                line=dict(color='#d62728', width=2, dash='dashdot'),
                marker=dict(size=8, symbol='x'),
                legendgroup="uniq_line"
            ),
            row=2, col=1
        )
        fig.update_yaxes(
            title_text="Uniqueness φ²",
            range=[-0.05, 1.10],
            row=2, col=1
        )
        fig.update_xaxes(
            title_text="Monitored Channels",
            tickangle=25,
            row=2, col=1
        )

        # ── (2,2) Latent Factor Variance Bar ───────────────────────────────────
        fig.add_trace(
            go.Bar(
                x=factor_labels,
                y=factor_variance,
                name="Factor Score Variance",
                marker=dict(
                    color='#2ca02c',
                    line=dict(color='black', width=0.5)
                ),
                legendgroup="fvar"
            ),
            row=2, col=2
        )
        fig.update_yaxes(title_text="Empirical Variance", row=2, col=2)
        fig.update_xaxes(title_text="Latent Factors",     row=2, col=2)

        # ── Global layout ──────────────────────────────────────────────────────
        fig.update_layout(
            title=dict(
                text="Factor Analysis (FA) — Latent Subspace Diagnostics Dashboard",
                x=0.5, y=0.97,
                xanchor="center", yanchor="top",
                font=dict(size=16)
            ),
            template="plotly_white",
            width=1250,
            height=750,
            showlegend=True,
            legend=dict(
                orientation="h",
                x=0.5,
                y=1.02,
                xanchor="center",
                yanchor="bottom"
            ),
            margin=dict(t=150, b=60, l=140, r=80)
        )

        fig.show()


print("FactorAnalysisEngine class defined successfully.")
````

````python
import numpy as np
import pandas as pd
from typing import Optional, Sequence, Dict, Any
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots


class DualSubspaceDiagnosticsEngine:
    """
    Unified engine implementing both PCA and FA decomposition pipelines
    with full Plotly diagnostic dashboards.
    """

    def __init__(self, df: pd.DataFrame):
        self.df = df

    # ──────────────────────────────────────────────────────────────────────────
    # PCA PIPELINE — 2 × 3 Optimization Dashboard
    # ──────────────────────────────────────────────────────────────────────────
    def compute_empirical_pca(
        self,
        columns: Optional[Sequence[str]] = None,
        show_plot: bool = True
    ) -> Dict[str, Any]:
        """
        Geometric de-correlation via PCA on the unbiased sample covariance S.
        Renders a 2 × 3 dashboard covering:
          - Loading matrix, eigenvalue bar chart, variance profile
          - Residual unexplained variance, Mean T², Mean Q (SPE)
        """
        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
        else:
            target_cols = list(columns)

        X    = self.df[target_cols].dropna().values
        n, m = X.shape

        if n <= m:
            raise ValueError("Require n > m to compute a full-rank covariance matrix.")

        # ── Centering & covariance ─────────────────────────────────────────────
        mu_hat     = X.mean(axis=0)
        X_centered = X - mu_hat
        S_matrix   = np.cov(X, rowvar=False, ddof=1)

        # ── Spectral decomposition ─────────────────────────────────────────────
        raw_vals, raw_vecs = np.linalg.eigh(S_matrix)
        order      = np.argsort(raw_vals)[::-1]
        lambda_hat = np.clip(raw_vals[order], 1e-15, None)
        P_hat      = raw_vecs[:, order]

        # ── Variance accounting ────────────────────────────────────────────────
        total_var          = lambda_hat.sum()
        expl_ratio         = lambda_hat / total_var
        cumul_ratio        = np.cumsum(expl_ratio)
        unexpl_ratio       = 1.0 - cumul_ratio

        # ── PC scores ─────────────────────────────────────────────────────────
        Z_scores = X_centered @ P_hat    # (n, m)
        S_Z      = np.cov(Z_scores, rowvar=False, ddof=1)

        # ── T² and Q vs k ──────────────────────────────────────────────────────
        k_range        = np.arange(1, m)
        mean_T2_vs_k   = []
        mean_Q_vs_k    = []
        T2_all         = np.zeros((n, len(k_range)))
        Q_all          = np.zeros((n, len(k_range)))

        for idx_k, k in enumerate(k_range):
            T2_i = np.sum((Z_scores[:, :k] ** 2) / lambda_hat[:k], axis=1)
            Q_i  = np.sum(Z_scores[:, k:] ** 2, axis=1)
            T2_all[:, idx_k] = T2_i
            Q_all[:, idx_k]  = Q_i
            mean_T2_vs_k.append(T2_i.mean())
            mean_Q_vs_k.append(Q_i.mean())

        print(f"\n─── PCA Engine ──────────────────────────────────────")
        print(f"  Features : {m}   |   Samples : {n}")
        print(f"  Trace(S) : {total_var:.4f}")

        # ── Dashboard ─────────────────────────────────────────────────────────
        if show_plot:
            pc_labels  = [f"PC {i+1}" for i in range(m)]
            k_labels   = [f"k={k}"  for k in k_range]

            fig = make_subplots(
                rows=2, cols=3,
                horizontal_spacing=0.18, vertical_spacing=0.28,
                subplot_titles=(
                    "Feature Loading Matrix |P̂|",
                    "Component Eigenvalues λ",
                    "Variance Profile (Explained)",
                    "Residual Space (Unexplained)",
                    "Mean Hotelling T² vs k",
                    "Mean Q Statistic (SPE) vs k"
                )
            )

            # (1,1) Loading heatmap
            fig.add_trace(go.Heatmap(
                z=np.abs(P_hat), x=pc_labels, y=target_cols,
                colorscale='YlOrRd',
                colorbar=dict(title="Weight", x=-0.12, len=0.38,
                              y=0.78, yanchor="middle", xanchor="right"),
                showscale=True, showlegend=False
            ), row=1, col=1)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=1)

            # (1,2) Eigenvalues
            fig.add_trace(go.Bar(
                x=pc_labels, y=lambda_hat, name="λ_j",
                marker=dict(color='#1f77b4', line=dict(color='black', width=0.5))
            ), row=1, col=2)
            fig.update_yaxes(title_text="Variance Magnitude", row=1, col=2)

            # (1,3) Marginal + cumulative explained variance
            fig.add_trace(go.Bar(
                x=pc_labels, y=expl_ratio * 100, name="Marginal %",
                marker=dict(color='#ff7f0e', opacity=0.75)
            ), row=1, col=3)
            fig.add_trace(go.Scatter(
                x=pc_labels, y=cumul_ratio * 100, mode='lines+markers',
                name='Cumulative %',
                line=dict(color='#d62728', width=2.5, dash='dash')
            ), row=1, col=3)
            fig.update_yaxes(title_text="Captured Variance (%)", range=[-2, 105], row=1, col=3)

            # (2,1) Residual unexplained
            fig.add_trace(go.Bar(
                x=pc_labels, y=unexpl_ratio * 100, name="Unexplained %",
                marker=dict(color='#2ca02c', line=dict(color='black', width=0.5))
            ), row=2, col=1)
            fig.update_yaxes(title_text="Excluded Info (%)", range=[-2, 105], row=2, col=1)

            # (2,2) Mean T²
            fig.add_trace(go.Scatter(
                x=k_labels, y=mean_T2_vs_k, mode='lines+markers',
                name='Mean T²',
                line=dict(color='#9467bd', width=2.5), marker=dict(symbol='diamond', size=6)
            ), row=2, col=2)
            fig.update_yaxes(title_text="Avg T² Score", row=2, col=2)
            fig.update_xaxes(title_text="Subspace Size k", row=2, col=2)

            # (2,3) Mean Q
            fig.add_trace(go.Scatter(
                x=k_labels, y=mean_Q_vs_k, mode='lines+markers',
                name='Mean Q (SPE)',
                line=dict(color='#e377c2', width=2.5), marker=dict(symbol='square', size=6)
            ), row=2, col=3)
            fig.update_yaxes(title_text="Avg Residual Energy", row=2, col=3)
            fig.update_xaxes(title_text="Truncation Cutoff k", row=2, col=3)

            fig.update_layout(
                title=dict(text="PCA — Optimization & Feature Loading Dashboard",
                           x=0.5, xanchor="center"),
                template="plotly_white", height=750, width=1250,
                legend=dict(orientation="h", y=1.02, x=0.5, xanchor="center"),
                margin=dict(t=150, b=60, l=140, r=80)
            )
            fig.show()

        return dict(
            mean_vector=mu_hat, covariance_S=S_matrix, eigenvalues=lambda_hat,
            eigenvectors_P=P_hat, explained_ratio=expl_ratio, cumulative_ratio=cumul_ratio,
            unexplained_ratio=unexpl_ratio, scores_Z=Z_scores,
            k_values=k_range, mean_T2_profile=np.array(mean_T2_vs_k),
            mean_Q_profile=np.array(mean_Q_vs_k)
        )

    # ──────────────────────────────────────────────────────────────────────────
    # FA PIPELINE — 2 × 2 Latent Subspace Dashboard
    # ──────────────────────────────────────────────────────────────────────────
    def compute_empirical_fa(
        self,
        k: int,
        columns: Optional[Sequence[str]] = None,
        show_plot: bool = True
    ) -> Dict[str, Any]:
        """
        Parametric FA with Varimax rotation. Renders a 2 × 2 dashboard.

        Guardrails:
          - Zero-variance channels are floored to 1e-15.
          - Requires n > m and k < m.
        """
        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
        else:
            target_cols = list(columns)

        X    = self.df[target_cols].dropna().values
        n, m = X.shape

        # Guardrails
        if n <= m:
            raise ValueError(f"Snapshot count n={n} must exceed feature count m={m}.")
        if k >= m:
            raise ValueError(f"Latent factor count k={k} must be strictly less than m={m}.")

        # ── Z-score standardisation ────────────────────────────────────────────
        mu_hat  = X.mean(axis=0)
        std_hat = np.std(X, axis=0, ddof=1)
        std_hat[std_hat == 0] = 1e-15        # zero-variance safeguard
        Z        = (X - mu_hat) / std_hat
        R_matrix = np.corrcoef(X, rowvar=False)

        # ── Fit FA with Varimax rotation ───────────────────────────────────────
        fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
        fa.fit(Z)
        lambda_matrix = fa.components_.T      # (m, k)
        psi_vec       = fa.noise_variance_    # uniqueness vector

        # ── Communality and uniqueness ─────────────────────────────────────────
        communality = (lambda_matrix ** 2).sum(axis=1)
        uniqueness  = psi_vec

        # ── Thomson's regression scores ────────────────────────────────────────
        F_scores = fa.transform(Z)              # (n, k)
        F_var    = np.var(F_scores, axis=0, ddof=1)

        print(f"\n─── FA Engine ───────────────────────────────────────")
        print(f"  Features m : {m}   |   Factors k : {k}   |   Samples n : {n}")
        print(f"  Avg Communality : {communality.mean()*100:.2f}%")
        print(f"  Avg Uniqueness  : {uniqueness.mean()*100:.2f}%")

        # ── 2 × 2 Dashboard ───────────────────────────────────────────────────
        if show_plot:
            factor_labels = [f"Factor {r+1}" for r in range(k)]

            fig = make_subplots(
                rows=2, cols=2,
                horizontal_spacing=0.24, vertical_spacing=0.28,
                subplot_titles=(
                    "Structural Loadings |Λ_rotated|",
                    "Variance Partition: Communality vs Uniqueness",
                    "Sensor Uniqueness Noise Floor (φ²)",
                    "Latent Factor Empirical Variance"
                )
            )

            # (1,1) Heatmap — left-anchored colorbar
            fig.add_trace(go.Heatmap(
                z=np.abs(lambda_matrix), x=factor_labels, y=target_cols,
                colorscale='YlOrRd',
                colorbar=dict(
                    title="Sensitivity", x=-0.15, len=0.38,
                    y=0.78, yanchor="middle", xanchor="right"
                ),
                showscale=True, showlegend=False
            ), row=1, col=1)
            fig.update_xaxes(title_text="Latent Structures", row=1, col=1)

            # (1,2) Stacked bar: communality + uniqueness = 100 %
            fig.add_trace(go.Bar(
                y=target_cols, x=communality * 100, orientation='h',
                name="Communality h² (Shared)",
                marker=dict(color='#1f77b4'), legendgroup="comm"
            ), row=1, col=2)
            fig.add_trace(go.Bar(
                y=target_cols, x=uniqueness * 100, orientation='h',
                name="Uniqueness φ² (Noise)",
                marker=dict(color='#ff7f0e'), legendgroup="uniq_bar"
            ), row=1, col=2)
            fig.update_layout(barmode='stack')
            fig.update_xaxes(title_text="Variance Allocation (%)", range=[0, 100], row=1, col=2)

            # (2,1) Uniqueness line profile
            fig.add_trace(go.Scatter(
                x=target_cols, y=uniqueness, mode='lines+markers',
                name="Uniqueness φ²",
                line=dict(color='#d62728', width=2, dash='dot'),
                marker=dict(size=8, symbol='x'), legendgroup="uniq_line"
            ), row=2, col=1)
            fig.update_yaxes(range=[-0.05, 1.05], row=2, col=1)
            fig.update_xaxes(title_text="Monitored Channels", tickangle=25, row=2, col=1)

            # (2,2) Factor variance bar
            fig.add_trace(go.Bar(
                x=factor_labels, y=F_var,
                name="Factor Variance",
                marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)),
                legendgroup="fvar"
            ), row=2, col=2)
            fig.update_yaxes(title_text="Variance Level", row=2, col=2)
            fig.update_xaxes(title_text="Latent Vectors", row=2, col=2)

            fig.update_layout(
                title=dict(text="FA — Latent Subspace Diagnostics Dashboard",
                           x=0.5, xanchor="center"),
                template="plotly_white", height=750, width=1250,
                legend=dict(orientation="h", y=1.02, x=0.5, xanchor="center"),
                margin=dict(t=150, b=60, l=140, r=80)
            )
            fig.show()

        return dict(
            mean_vector=mu_hat, std_vector=std_hat, correlation_R=R_matrix,
            loadings_lambda=lambda_matrix, uniqueness=uniqueness,
            communality=communality, factor_scores=F_scores, sensors=target_cols
        )


# ── Data Generator & Pipeline Execution ───────────────────────────────────────
np.random.seed(42)
n_samples = 2500

f1 = np.random.normal(0, 1, n_samples)
f2 = np.random.normal(0, 1, n_samples)

s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.30, n_samples)
s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)
s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)
s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)   # faulty sensor

df_asset = pd.DataFrame(
    np.vstack([s1, s2, s3, s4]).T,
    columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
)

engine = DualSubspaceDiagnosticsEngine(df_asset)
pca_metrics = engine.compute_empirical_pca(show_plot=True)
fa_metrics  = engine.compute_empirical_fa(k=2, show_plot=True)
````

### Solution for Question 1: The Total Variance Illusion



**What a uniqueness $\varphi^2 \approx 100\%$ means for Sensor 4:**

When virtually all of a sensor's variance is categorised as uniqueness, the channel contributes **nothing** to the shared structural signal. The factor model cannot extract any useful common information from it — its output is dominated by isolated electrical noise or an instrument fault that is entirely uncorrelated with the rest of the array.

**How Sensor 4's noise inflates PCA eigenvalues:**

PCA treats *all* variance as equally informative, regardless of origin. With $\sigma^2_{\text{S4}} \approx 2.0$ dwarfing the variance of the well-behaved sensors ($\sigma^2 \approx 1.0$), PCA's greedy first axis (`PC 1`) is almost entirely commandeered by Sensor 4's noise. The resulting eigenvalue $\lambda_1 \approx 2.0$ does not reflect any underlying structural mode — it is pure measurement artifact.

**Engineering risk:**

An engineer who deploys only a PCA monitoring chart would find their primary control chart ($T^2$ on `PC 1`) responding almost exclusively to the noise floor of a broken sensor rather than to genuine structural anomalies. A real fault on the structure (affecting Sensors 1–3) could easily go unnoticed because those signals are compressed into the lower-priority components.

### Solution for Question 2: Decoupling Structural Loading via Rotation



**Varimax "simple structure" vs. raw PCA eigenvectors:**

A raw PCA eigenvector aims to explain as much total variance as possible in a single pass, so each component ends up mixing signals from multiple sensors simultaneously. The second principal component, for example, may show moderate loadings on Sensors 1, 2, and 3 all at once — a blend that has no clean physical interpretation.

Varimax releases the strict variance-maximisation ordering and instead rotates the loading matrix until each column has a few very large entries and many near-zero ones (maximising the within-column variance of squared loadings). The result is that Sensors 1 and 2, which share the dominant physical process $f_1$, cluster together on Factor 1, while Sensor 3, driven by the independent process $f_2$, sits alone on Factor 2.

**Operational interpretability advantage:**

With the raw PCA structure, an alarm on `PC 2` forces the operator to mentally de-mix three sensors simultaneously. With the rotated FA structure, an alarm on `Factor 1` immediately points to the $f_1$-driven subsystem (Sensors 1 & 2), and an alarm on `Factor 2` points exclusively to Sensor 3. Diagnosis time drops from minutes to seconds.

### Solution for Question 3: Determining Subspace Truncation ($k$)



**Behaviour of the Mean $Q$ Statistic curve:**

- **$k = 1 \to k = 2$:** The curve falls steeply. Including a second component absorbs a large block of structured residual energy. The drop is pronounced because there are two genuinely distinct sources of correlated variance in the data.
- **$k = 2 \to k = 3$:** The curve flattens into a clear "elbow". Marginal gains become small, indicating that the third component explains only noise.

**Why the elbow identifies $k = 2$ as the true dimensionality:**

The $Q$ statistic measures unmodelled energy. A sharp initial drop followed by a plateau indicates that the structured information content has been fully captured at $k = 2$. Everything beyond that elbow is residual white noise that the monitoring system should *not* attempt to model.

Cross-referencing with the loading heatmap confirms the interpretation: `PC 1` maps to the noise-dominated `Sensor 4`, `PC 2` captures the correlated triad of Sensors 1–3, and `PC 3` onwards represents leftover noise.

**Consequence of choosing $k = 3$:**

Pulling the third component into the "clean" principal subspace imports random noise as if it were structured signal. This overfits the model, inflating the control limits for $T^2$ and generating frequent spurious alarms whenever normal baseline noise temporarily pushes a sample across the now-too-tight boundary.

### Solution for Question 4: Operational Trade-offs in System Health Monitoring



#### PCA Strategy ($T^2 + Q$ on principal components)

**Strengths:** No parametric model assumptions required; straightforward to implement; $Q$ is an excellent detector of novel fault directions.

**Weakness — sensor degradation vulnerability:** PCA principal axes are a global function of the full covariance matrix. If one sensor malfunctions (e.g., `Sensor 4` shorts and its noise variance doubles), the eigenvectors rotate to accommodate the new covariance structure. All control limits and $T^2$ scores shift, even for perfectly healthy sensors. The monitoring system effectively loses calibration across the board.

#### FA Strategy (Rotated Latent Factor Scores)

**Strengths:** The diagonal uniqueness matrix $\boldsymbol{\Psi}$ explicitly quarantines sensor-specific noise. If `Sensor 4` degrades further, its uniqueness $\varphi_4^2 \to 1$, but this is absorbed into the noise floor without contaminating the latent factor estimates. The Factor 1 and Factor 2 scores remain stable and trustworthy.

**Conclusion:** The FA strategy is significantly more robust against single-sensor calibration loss or electrical faults. By explicitly separating shared structural variance from localized channel noise, it protects the diagnostic layer from being hijacked by one malfunctioning instrument — a critical property in any large-scale industrial monitoring deployment.

| Criterion | PCA + $T^2/Q$ | Rotated FA |
|---|---|---|
| Single sensor failure | Global axis rotation — all limits shift | Absorbed by $\varphi^2$; factors unaffected |
| Novel fault direction | Excellent ($Q$ captures it) | Moderate (falls in residual space) |
| Physical interpretability | Limited | High (each factor maps to a process) |
| Model assumptions | Non-parametric | Multivariate Normal required |
