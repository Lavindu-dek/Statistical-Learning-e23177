# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

## Statistical Mechanics of the Quantum Harmonic Oscillator

### 1. The Partition Function

Starting from the definition, with $\beta \equiv 1/(k_BT)$:

$$Z(T) = \sum_{j=0}^\infty e^{-\beta E_j} = \sum_{j=0}^\infty e^{-\beta\hbar\omega(j+1/2)}$$

Pull out the $j$-independent piece:

$$Z(T) = e^{-\beta\hbar\omega/2}\sum_{j=0}^\infty e^{-\beta\hbar\omega j} = e^{-\beta\hbar\omega/2}\sum_{j=0}^\infty \left(e^{-\beta\hbar\omega}\right)^j$$

This is a geometric series with common ratio $r = e^{-\beta\hbar\omega}$. Since $\beta\hbar\omega > 0$ for any physical temperature, we have $0 < r < 1$, guaranteeing convergence. Using $\sum_{j=0}^\infty r^j = \dfrac{1}{1-r}$:

$$Z(T) = \frac{e^{-\beta\hbar\omega/2}}{1-e^{-\beta\hbar\omega}} = \frac{e^{-\frac{\hbar\omega}{2k_BT}}}{1-e^{-\frac{\hbar\omega}{k_BT}}}$$

---

### 2. Thermodynamic Potentials

**Helmholtz Free Energy**

$$F = -k_BT\ln Z = -k_BT\left[-\frac{\beta\hbar\omega}{2} - \ln\left(1-e^{-\beta\hbar\omega}\right)\right]$$

The first term simplifies since $-k_BT \cdot(-\beta\hbar\omega/2) = \hbar\omega/2$ (because $k_BT\beta = 1$):

$$F(T) = \frac{\hbar\omega}{2} + k_BT\ln\left(1-e^{-\hbar\omega/k_BT}\right)$$

**Internal Energy**

Using $\ln Z = -\dfrac{\beta\hbar\omega}{2} - \ln(1-e^{-\beta\hbar\omega})$:

$$\frac{\partial \ln Z}{\partial \beta} = -\frac{\hbar\omega}{2} - \frac{\hbar\omega\, e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}}$$

So:

$$U = -\frac{\partial \ln Z}{\partial \beta} = \frac{\hbar\omega}{2} + \frac{\hbar\omega\, e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}}$$

Multiplying numerator and denominator by $e^{\beta\hbar\omega}$:

$$U(T) = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\hbar\omega/k_BT}-1}$$

This is the familiar zero-point energy plus the Planck (Bose–Einstein) occupation term.

---

### 3. Entropy

Compute $S = -\partial F/\partial T$. Let $\theta \equiv \hbar\omega/k_B$, so $F = \dfrac{\hbar\omega}{2} + k_BT\ln(1-e^{-\theta/T})$.

**Product rule** on the second term:

$$\frac{\partial F}{\partial T} = k_B\ln\left(1-e^{-\theta/T}\right) + k_BT \cdot \frac{\partial}{\partial T}\ln\left(1-e^{-\theta/T}\right)$$

Since $\dfrac{\partial}{\partial T}e^{-\theta/T} = e^{-\theta/T}\cdot\dfrac{\theta}{T^2}$:

$$\frac{\partial}{\partial T}\ln\left(1-e^{-\theta/T}\right) = \frac{e^{-\theta/T}\,\theta/T^2}{1-e^{-\theta/T}}$$

Therefore:

$$\frac{\partial F}{\partial T} = k_B\ln\left(1-e^{-\theta/T}\right) + \frac{k_B\theta}{T}\cdot\frac{e^{-\theta/T}}{1-e^{-\theta/T}}$$

So, restoring $\theta/T = \beta\hbar\omega$:

$$S(T) = k_B\left[\frac{\beta\hbar\omega\, e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}} - \ln\left(1-e^{-\beta\hbar\omega}\right)\right]$$

**Verification: $U = F + TS$**

$$F + TS = \left[\frac{\hbar\omega}{2} + k_BT\ln(1-e^{-\beta\hbar\omega})\right] + T\cdot k_B\left[\frac{\beta\hbar\omega\,e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}} - \ln(1-e^{-\beta\hbar\omega})\right]$$

The $k_BT\ln(1-e^{-\beta\hbar\omega})$ terms cancel exactly, leaving:

$$F+TS = \frac{\hbar\omega}{2} + Tk_B\beta\hbar\omega\cdot\frac{e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}}$$

Since $Tk_B\beta = 1$:

$$F+TS = \frac{\hbar\omega}{2} + \frac{\hbar\omega\,e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}} = U \;\checkmark$$

confirming consistency with the $U(T)$ found in Part 2.

---

### 4. Heat Capacity and Hyperbolic Form

**Direct differentiation**

Starting from $U(T) = \dfrac{\hbar\omega}{2} + \hbar\omega\left(e^{\hbar\omega/k_BT}-1\right)^{-1}$, let $y = \hbar\omega/(k_BT)$, so $dy/dT = -\hbar\omega/(k_BT^2)$.

$$\frac{\partial U}{\partial T} = \hbar\omega\left(-1\right)\frac{e^y}{(e^y-1)^2}\left(-\frac{\hbar\omega}{k_BT^2}\right)$$

$$C_V = \frac{\hbar^2\omega^2}{k_BT^2}\cdot\frac{e^{\hbar\omega/k_BT}}{\left(e^{\hbar\omega/k_BT}-1\right)^2} = k_B\left(\frac{\hbar\omega}{k_BT}\right)^2\frac{e^{\hbar\omega/k_BT}}{\left(e^{\hbar\omega/k_BT}-1\right)^2}$$

**Converting to hyperbolic form**

Let $x \equiv \dfrac{\hbar\omega}{2k_BT}$, so $\dfrac{\hbar\omega}{k_BT} = 2x$. Factor the denominator:

$$e^{2x}-1 = e^{x}\left(e^{x}-e^{-x}\right) = e^x\cdot 2\sinh(x)$$

using $\sinh(x) = \dfrac{e^x-e^{-x}}{2}$. Therefore:

$$\left(e^{2x}-1\right)^2 = e^{2x}\cdot 4\sinh^2(x)$$

Substituting back:

$$C_V = k_B(2x)^2 \cdot \frac{e^{2x}}{e^{2x}\cdot 4\sinh^2(x)} = k_B\,x^2\,\text{csch}^2(x)$$

The exponential factors cancel cleanly, leaving:

$$C_V = k_B\left(\frac{\hbar\omega}{2k_BT}\right)^2\frac{1}{\sinh^2\left(\dfrac{\hbar\omega}{2k_BT}\right)}$$

as required.

---

### 5. High-Temperature Asymptotics

In the limit $k_BT \gg \hbar\omega$, the argument $x = \dfrac{\hbar\omega}{2k_BT} \to 0$. The Taylor expansion of $\sinh(x)$ is:

$$\sinh(x) = x + \frac{x^3}{6} + O(x^5) \;\approx\; x \quad (x\to 0)$$

so $\sinh^2(x) \approx x^2$, and:

$$C_V \approx k_B\cdot x^2 \cdot \frac{1}{x^2} = k_B$$

**Result:** $C_V \to k_B$ as $T\to\infty$ (or more precisely, whenever $k_BT \gg \hbar\omega$).

**Comparison with equipartition.** The classical equipartition theorem assigns $\tfrac12 k_B$ per quadratic degree of freedom. A classical harmonic oscillator has *two* quadratic terms — kinetic energy $p^2/2m$ and potential energy $\tfrac12 m\omega^2 x^2$ — giving:

$$C_V^{classical} = \frac{1}{2}k_B + \frac{1}{2}k_B = k_B$$

This matches the high-temperature quantum result exactly — the expected **correspondence principle** behavior: when thermal energy $k_BT$ greatly exceeds the level spacing $\hbar\omega$, the discreteness of the quantum spectrum becomes negligible and the system behaves classically. Conversely, at low temperature ($k_BT \ll \hbar\omega$, $x\to\infty$), $\sinh^2(x)$ grows exponentially, so $C_V \to 0$ exponentially fast — a purely quantum effect (the oscillator "freezes out" into its ground state) with no classical analogue, resolving the historical puzzle of vanishing heat capacities at low temperature.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

## The Maximum Entropy Distribution under Mean and Variance Constraints

### Setup

We seek the density $p(x)$ on $x \in (-\infty, \infty)$ that maximizes the differential (continuous) entropy

$$H[p] = -\int_{-\infty}^{\infty} p(x)\ln p(x)\,dx$$

subject to three constraints:

$$\text{Normalization:}\quad \int_{-\infty}^{\infty} p(x)\,dx = 1$$

$$\text{Mean:}\quad \int_{-\infty}^{\infty} x\,p(x)\,dx = \mu$$

$$\text{Variance:}\quad \int_{-\infty}^{\infty} (x-\mu)^2\,p(x)\,dx = \sigma^2$$

### Step 1: Form the Lagrangian

Attach a multiplier $\gamma$ to normalization, $\lambda_1$ to the mean constraint, and $\lambda_2$ to the variance constraint. Since we are optimizing over a *function* $p(x)$ rather than a finite list of variables, this is a calculus-of-variations problem. The Lagrangian functional is:

$$\mathscr{L}[p] = -\int p(x)\ln p(x)\,dx \;-\; \gamma\left(\int p(x)\,dx - 1\right) \;-\; \lambda_1\left(\int x\,p(x)\,dx - \mu\right) \;-\; \lambda_2\left(\int (x-\mu)^2 p(x)\,dx - \sigma^2\right)$$

We can write this entirely as a single integral of a functional density $\mathscr{L}[p] = \int \ell\big(p(x), x\big)\,dx + \text{const}$, where

$$\ell\big(p(x),x\big) = -p(x)\ln p(x) - \gamma\, p(x) - \lambda_1\, x\, p(x) - \lambda_2\,(x-\mu)^2 p(x)$$

### Step 2: Apply the Euler–Lagrange Condition

Since $\ell$ depends on $p(x)$ pointwise (no derivatives of $p$ appear), the stationarity condition reduces to an ordinary pointwise derivative with respect to $p(x)$, treated as a free variable at each $x$:

$$\frac{\partial \ell}{\partial p(x)} = 0$$

Differentiating term by term:

$$\frac{\partial}{\partial p}\Big[-p\ln p\Big] = -\ln p - 1$$

$$\frac{\partial}{\partial p}\Big[-\gamma p\Big] = -\gamma, \qquad \frac{\partial}{\partial p}\Big[-\lambda_1 x p\Big] = -\lambda_1 x, \qquad \frac{\partial}{\partial p}\Big[-\lambda_2(x-\mu)^2 p\Big] = -\lambda_2(x-\mu)^2$$

Setting the sum to zero:

$$-\ln p(x) - 1 - \gamma - \lambda_1 x - \lambda_2(x-\mu)^2 = 0$$

Solving for $p(x)$:

$$p(x) = \exp\Big(-1-\gamma - \lambda_1 x - \lambda_2(x-\mu)^2\Big)$$

This confirms the general MaxEnt template: an exponential family with a linear term (from the mean constraint) and a quadratic term (from the variance constraint). The presence of the quadratic penalty $-\lambda_2(x-\mu)^2$ already tells us the shape must be Gaussian — we now pin down the constants.

### Step 3: Absorb Constants and Complete the Square

Group the constant prefactor as $A \equiv e^{-1-\gamma}$:

$$p(x) = A\exp\Big(-\lambda_1 x - \lambda_2(x-\mu)^2\Big)$$

To see this is a shifted Gaussian, expand the exponent and complete the square in $x$. Write $u = x - \mu$, so $x = u+\mu$:

$$-\lambda_1(u+\mu) - \lambda_2 u^2 = -\lambda_2 u^2 - \lambda_1 u - \lambda_1\mu$$

Complete the square in $u$:

$$-\lambda_2\left(u^2 + \frac{\lambda_1}{\lambda_2}u\right) - \lambda_1\mu = -\lambda_2\left(u + \frac{\lambda_1}{2\lambda_2}\right)^2 + \frac{\lambda_1^2}{4\lambda_2} - \lambda_1\mu$$

So:

$$p(x) = A\, e^{\frac{\lambda_1^2}{4\lambda_2}-\lambda_1\mu}\,\exp\left(-\lambda_2\left(u+\frac{\lambda_1}{2\lambda_2}\right)^2\right)$$

This is a Gaussian in $u = x-\mu$ centered at $u^\* = -\lambda_1/(2\lambda_2)$, i.e., centered in $x$ at $x^\* = \mu - \lambda_1/(2\lambda_2)$.

### Step 4: Use the Mean Constraint to Force $\lambda_1 = 0$

A distribution that is symmetric about its center automatically has that center equal to its mean. We are told the mean is exactly $\mu$. If $\lambda_1 \ne 0$, the peak (and by symmetry, the mean) of this Gaussian would sit at $x^\* = \mu - \lambda_1/(2\lambda_2) \ne \mu$, contradicting the constraint $\mathbb{E}[X]=\mu$.

More rigorously: since $p(x)$ has the form $A'\exp(-\lambda_2(x-\mu)^2 - \lambda_1(x-\mu))$ (absorbing constants), the mean of this shifted-and-tilted Gaussian is $\mu - \lambda_1/(2\lambda_2)$. Setting this equal to $\mu$ gives immediately:

$$\lambda_1 = 0$$

With $\lambda_1 = 0$, the density collapses to a clean, symmetric Gaussian centered exactly at $\mu$:

$$p(x) = A\,e^{0}\exp\big(-\lambda_2(x-\mu)^2\big) = A\exp\big(-\lambda_2(x-\mu)^2\big)$$

### Step 5: Use Normalization to Fix $A$ in Terms of $\lambda_2$

We need $\int_{-\infty}^{\infty} A\, e^{-\lambda_2(x-\mu)^2}\,dx = 1$. Using the standard Gaussian integral $\displaystyle\int_{-\infty}^{\infty} e^{-a t^2}\,dt = \sqrt{\pi/a}$ (valid for $\lambda_2>0$, which we require for the integral to converge — the exponent must decay, not blow up):

$$A\sqrt{\frac{\pi}{\lambda_2}} = 1 \quad\Longrightarrow\quad A = \sqrt{\frac{\lambda_2}{\pi}}$$

So the density is now:

$$p(x) = \sqrt{\frac{\lambda_2}{\pi}}\,\exp\big(-\lambda_2(x-\mu)^2\big)$$

### Step 6: Use the Variance Constraint to Fix $\lambda_2$

We need $\displaystyle\int_{-\infty}^{\infty} (x-\mu)^2\, p(x)\,dx = \sigma^2$. This is the second moment of a centered Gaussian with density $\sqrt{\lambda_2/\pi}\,e^{-\lambda_2 t^2}$ (with $t = x-\mu$). Using the standard result

$$\int_{-\infty}^{\infty} t^2\, e^{-a t^2}\,dt = \frac{1}{2a}\sqrt{\frac{\pi}{a}}$$

with $a = \lambda_2$:

$$\int_{-\infty}^{\infty} t^2\sqrt{\frac{\lambda_2}{\pi}}e^{-\lambda_2 t^2}\,dt = \sqrt{\frac{\lambda_2}{\pi}}\cdot\frac{1}{2\lambda_2}\sqrt{\frac{\pi}{\lambda_2}} = \frac{1}{2\lambda_2}$$

Setting this equal to $\sigma^2$:

$$\frac{1}{2\lambda_2} = \sigma^2 \quad\Longrightarrow\quad \lambda_2 = \frac{1}{2\sigma^2}$$

### Step 7: Assemble the Final Distribution

Substituting $\lambda_2 = 1/(2\sigma^2)$ back into both $A$ and the exponent:

$$A = \sqrt{\frac{\lambda_2}{\pi}} = \sqrt{\frac{1}{2\pi\sigma^2}} = \frac{1}{\sqrt{2\pi\sigma^2}}$$

$$p(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

which is exactly the target result:

$$p^\*(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

### Interpretation

This is the central result underpinning the ubiquity of the Gaussian/Normal distribution: **among all continuous distributions on $(-\infty,\infty)$ with a specified mean and variance, the Normal distribution is the unique one that is maximally uncertain (least informative) beyond those two facts.** Any other distribution matching the same mean and variance necessarily encodes additional structure (skewness, extra peaks, heavier or lighter tails) not implied by the mean/variance data alone — and so has strictly lower entropy than the Gaussian. This is part of why the Gaussian is the natural "default" or "least-biased" choice whenever only first and second moments are known — a fact that also underlies its central role in the Central Limit Theorem and in statistical thermodynamics (e.g., the Maxwell–Boltzmann velocity distribution).

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).




## Multiclass Logistic Regression: Automatic Classification of Emails

### Setup

An e-commerce platform routes support emails to one of two classes: $Y=0$ (Finance) or $Y=1$ (Technical), based on a binary feature vector

$$\mathbf{x} = (x_1, x_2, x_3)^T$$

where $x_1$ flags financial context, $x_2$ flags account-access context, and $x_3$ flags bug/malfunction context.

**Observed scenario probabilities** (marginal distribution over $\mathbf{x}$):

| Scenario | $\mathbf{x}$ | $P(\mathbf{x})$ |
|---|---|---|
| A: Pure Financial | $(1,0,0)$ | 0.40 |
| B: Pure Access | $(0,1,0)$ | 0.25 |
| C: Pure Technical | $(0,0,1)$ | 0.05 |
| D: Access + Technical | $(0,1,1)$ | 0.15 |
| E: Financial + Access | $(1,1,0)$ | 0.05 |

**Audited joint constraints** on the label:

$$\mathbb{E}[x_1\,\mathbb{I}(Y=1)] = 0.05, \qquad \mathbb{E}[x_2\,\mathbb{I}(Y=1)] = 0.15, \qquad \mathbb{E}[x_3\,\mathbb{I}(Y=1)] = 0.16$$

### The MaxEnt / Softmax Model

Using the standard uniform indicator feature mapping $\mathbf{f}(\mathbf{x},c) = \mathbb{I}(Y=c)\,\mathbf{x}$, the MaxEnt conditional distribution consistent with linear constraints on $\mathbb{E}[x_j\,\mathbb{I}(Y=c)]$ is the multiclass **Softmax** model, parameterized by weight vectors $\mathbf{w}_0, \mathbf{w}_1 \in \mathbb{R}^3$:

$$P(Y=c\mid\mathbf{x}) = \frac{\exp(\mathbf{w}_c\cdot\mathbf{x})}{\sum_{l=0}^{1}\exp(\mathbf{w}_l\cdot\mathbf{x})} = \frac{e^{\mathbf{w}_c\cdot\mathbf{x}}}{e^{\mathbf{w}_0\cdot\mathbf{x}}+e^{\mathbf{w}_1\cdot\mathbf{x}}}$$

With only two classes, this collapses to a **logistic (sigmoid)** form. Defining the difference vector $\mathbf{u} = \mathbf{w}_1-\mathbf{w}_0$ (so $u_j = w_{1j}-w_{0j}$):

$$P(Y=0\mid\mathbf{x}) = \frac{1}{1+e^{\mathbf{u}\cdot\mathbf{x}}}, \qquad P(Y=1\mid\mathbf{x}) = \frac{1}{1+e^{-\mathbf{u}\cdot\mathbf{x}}} = \sigma(\mathbf{u}\cdot\mathbf{x})$$

where $\sigma(\cdot)$ is the logistic sigmoid. Only the **difference** $\mathbf{u}$ is identifiable; the individual $\mathbf{w}_0,\mathbf{w}_1$ are not, since only their difference enters the ratio.

### Expanding the Moment Constraints

Because $P(Y,\mathbf{x}) = P(Y\mid\mathbf{x})\,P(\mathbf{x})$, each audited joint expectation decomposes as a sum over the five observed scenarios in $\mathscr{X}$:

$$\mathbb{E}[x_j\,\mathbb{I}(Y=1)] = \sum_{\mathbf{x}\in\mathscr{X}} x_j\cdot P(Y=1\mid\mathbf{x})\cdot P(\mathbf{x})$$

**For $j=1$ (Financial constraint).** Only Scenarios A and E have $x_1=1$:

$$\mathbb{E}[x_1\mathbb{I}(Y=1)] = 0.40\,\sigma(u_1) + 0.05\,\sigma(u_1+u_2)$$

**For $j=2$ (Access constraint).** Scenarios B, D, and E have $x_2=1$:

$$\mathbb{E}[x_2\mathbb{I}(Y=1)] = 0.25\,\sigma(u_2) + 0.15\,\sigma(u_2+u_3) + 0.05\,\sigma(u_1+u_2)$$

**For $j=3$ (Technical constraint).** Scenarios C and D have $x_3=1$:

$$\mathbb{E}[x_3\mathbb{I}(Y=1)] = 0.05\,\sigma(u_3) + 0.15\,\sigma(u_2+u_3)$$

Substituting the audited targets gives the system:

$$0.05 = 0.40\,\sigma(u_1) + 0.05\,\sigma(u_1+u_2) \tag{1}$$

$$0.15 = 0.25\,\sigma(u_2) + 0.15\,\sigma(u_2+u_3) + 0.05\,\sigma(u_1+u_2) \tag{2}$$

$$0.16 = 0.05\,\sigma(u_3) + 0.15\,\sigma(u_2+u_3) \tag{3}$$

### Step 1: Feasibility Bound on Equation (3)

Since $\sigma(\cdot) \in (0,1)$ strictly, the right-hand side of (3) satisfies

$$0.05\,\sigma(u_3) + 0.15\,\sigma(u_2+u_3) \;<\; 0.05(1) + 0.15(1) = 0.20$$

Since the target $0.16 < 0.20$, equation (3) is feasible. Isolating the Scenario-D term:

$$\sigma(u_2+u_3) = \frac{0.16 - 0.05\,\sigma(u_3)}{0.15} = 1.0667 - 0.3333\,\sigma(u_3)$$

### Step 2: Feasibility Bound on Equation (1)

Isolating the Scenario-E term in (1):

$$0.05\,\sigma(u_1+u_2) = 0.05 - 0.40\,\sigma(u_1) \quad\Longrightarrow\quad \sigma(u_1+u_2) = 1 - 8\,\sigma(u_1)$$

Since $\sigma(u_1+u_2)\ge 0$, this forces

$$\sigma(u_1) \le \frac{1}{8} = 0.125$$

meaning $u_1$ must be substantially negative — financial-context emails strongly repel Technical routing.

### Step 3: Combine into a Single Equation

Substitute both isolated expressions into equation (2):

$$0.15 = 0.25\,\sigma(u_2) + 0.15\Big[1.0667 - 0.3333\,\sigma(u_3)\Big] + 0.05\Big[1 - 8\,\sigma(u_1)\Big]$$

Expanding:

$$0.15 = 0.25\,\sigma(u_2) + 0.16 - 0.05\,\sigma(u_3) + 0.05 - 0.40\,\sigma(u_1)$$

$$-0.06 = 0.25\,\sigma(u_2) - 0.05\,\sigma(u_3) - 0.40\,\sigma(u_1)$$

This, together with the two relations from Steps 1–2, forms a fully determined (if coupled/nonlinear) system in $(u_1,u_2,u_3)$.

### Step 4: Numerical Solution

Solving the system numerically (e.g., Newton-Raphson or a root-finder) yields:

$$u_1 \approx -2.4845, \qquad u_2 \approx 1.0421, \qquad u_3 \approx 2.3369$$

Substituting into the sigmoidal model:

$$P(Y=1\mid\mathbf{x}) = \frac{1}{1+e^{-(-2.4845)x_1 - (1.0421)x_2 - (2.3369)x_3}} = \frac{1}{1+e^{2.4845x_1 - 1.0421x_2 - 2.3369x_3}}$$

The signs are intuitive: financial keywords ($u_1 \ll 0$) push strongly *away* from Technical routing, technical keywords ($u_3 \gg 0$) push strongly *toward* it, and access keywords ($u_2 > 0$) push moderately toward it.

### Step 5: Evaluate on Each Scenario

**Scenario A (Pure Financial):** $\mathbf{x}_A=(1,0,0)$

$$P(Y=1\mid\mathbf{x}_A) = \frac{1}{1+e^{2.4845}} \approx \frac{1}{1+11.995} \approx 0.0769$$

**Scenario B (Pure Access):** $\mathbf{x}_B=(0,1,0)$

$$P(Y=1\mid\mathbf{x}_B) = \frac{1}{1+e^{-1.0421}} \approx \frac{1}{1+0.3527} \approx 0.7393$$

**Scenario C (Pure Technical):** $\mathbf{x}_C=(0,0,1)$

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



## Asset Pricing and Portfolio Optimization: Risk-Neutral Density via MaxEnt

### Background

Asset X trades at \$100 today. We want the risk-neutral density $p(S_T)$ governing its price one month from now, $S_T \in (0,\infty)$, using only what the options market tells us — without imposing a Gaussian shape by assumption.

**Market-implied constraints:**

$$\mathbb{E}[S_T] = 102 \qquad \text{(forward price)}$$

$$\mathbb{E}[S_T^2] = 10{,}600 \qquad \text{(volatility-implied second moment)}$$

$$\mathbb{E}[\ln S_T] = 4.60 \qquad \text{(skew / crash-risk penalty)}$$

### 1. The Optimization Problem

We maximize the continuous (Shannon–Jaynes) entropy functional, using a uniform reference measure $m(S_T)=1$ since prices are strictly positive:

$$\max_{p}\; H_c[p] = -\int_0^\infty p(S_T)\ln p(S_T)\,dS_T$$

subject to:

| Constraint | Equation | Multiplier |
|---|---|---|
| Normalization | $\int_0^\infty p(S_T)\,dS_T = 1$ | $\gamma$ |
| Mean | $\int_0^\infty p(S_T)\,S_T\,dS_T = 102$ | $\lambda_1$ |
| Second moment | $\int_0^\infty p(S_T)\,S_T^2\,dS_T = 10{,}600$ | $\lambda_2$ |
| Skew | $\int_0^\infty p(S_T)\,\ln(S_T)\,dS_T = 4.60$ | $\lambda_3$ |

### 2. The MaxEnt Solution Structure

Following the same Lagrangian calculus as the general derivation (Euler–Lagrange stationarity in $p$), the constrained optimum takes the exponential-family form:

$$p^\*(S_T) = \frac{1}{Z}\exp\Big(-\lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3\ln(S_T)\Big)$$

Using the algebraic identity $\exp(-\lambda_3\ln S_T) = S_T^{-\lambda_3}$, this can be rewritten as a Gaussian-like kernel reshaped by a power-law factor:

$$p^\*(S_T) = \frac{1}{Z}\,S_T^{-\lambda_3}\,\exp\Big(-\lambda_1 S_T - \lambda_2 S_T^2\Big)$$

where the partition function normalizes the density:

$$Z(\lambda_1,\lambda_2,\lambda_3) = \int_0^\infty S_T^{-\lambda_3}\exp\Big(-\lambda_1 S_T - \lambda_2 S_T^2\Big)\,dS_T$$

Note that if $\lambda_3=0$ this reduces to the ordinary Gaussian case; the log-moment constraint is precisely what introduces the extra power-law distortion needed to model skewed, fat-tailed behavior.

### 3. Solving for the Multipliers

As in the general MaxEnt framework, each constraint's expectation is obtained by differentiating $\ln Z$ with respect to its multiplier:

**Mean constraint:**

$$-\frac{\partial \ln Z}{\partial \lambda_1} = 102 \quad\Longrightarrow\quad \frac{1}{Z}\int_0^\infty S_T^{1-\lambda_3}\exp\big(-\lambda_1 S_T - \lambda_2 S_T^2\big)\,dS_T = 102$$

**Second-moment constraint:**

$$-\frac{\partial \ln Z}{\partial \lambda_2} = 10{,}600 \quad\Longrightarrow\quad \frac{1}{Z}\int_0^\infty S_T^{2-\lambda_3}\exp\big(-\lambda_1 S_T - \lambda_2 S_T^2\big)\,dS_T = 10{,}600$$

**Skew constraint:**

$$-\frac{\partial \ln Z}{\partial \lambda_3} = 4.60 \quad\Longrightarrow\quad \frac{1}{Z}\int_0^\infty \ln(S_T)\,S_T^{-\lambda_3}\exp\big(-\lambda_1 S_T - \lambda_2 S_T^2\big)\,dS_T = 4.60$$

### 4. Numerical Root-Finding

Unlike the pure Gaussian case (where $\lambda_3=0$ and the multipliers have closed-form solutions via completing the square), these three integral equations cannot be isolated algebraically because of the coupled power-law and Gaussian-type factors. Quantitative analysts instead minimize the dual objective — the log-partition function plus the linear constraint terms — via a Newton–Raphson or quasi-Newton (BFGS) routine.

Solving numerically for this market data yields:

$$\lambda_1 \approx -0.1843, \qquad \lambda_2 \approx 0.00114, \qquad \lambda_3 \approx 0.4215, \qquad Z \approx 134.72$$

**Interpreting each multiplier:**

- $\lambda_1 < 0$: acts as a linear growth driver, shifting probability mass forward toward the \$102 risk-free forward price.
- $\lambda_2 > 0$: the quadratic term provides tail-decay, preventing the price distribution from diverging at large $S_T$.
- $\lambda_3 > 0$: since $\exp(-\lambda_3\ln S_T) = S_T^{-0.4215}$, this creates a strong dampening effect as $S_T\to 0$, hollowing out probability mass near zero and producing a steep, fat left tail — precisely the crash-risk skew priced into the options market.

### 5. The Final Probability Density

Substituting the solved parameters gives the completely unbiased (maximum entropy), market-consistent asset price distribution:

$$p^\*(S_T) = \frac{1}{134.72}\cdot\frac{\exp\big(0.1843\,S_T - 0.00114\,S_T^2\big)}{S_T^{0.4215}}, \qquad S_T > 0$$

### Interpretation

This distribution is the least-biased choice consistent with everything the market tells us — the forward price, the implied variance, and the skew from put-option trading — without assuming Gaussian symmetry. Compared to a Normal distribution matched only to the same mean and variance, this MaxEnt density is asymmetric: it decays more sharply near $S_T=0$ (reflecting the market's aversion to prices collapsing) while permitting a heavier right-leaning bulk consistent with the forward price target. This flexibility is exactly why MaxEnt is valuable for pricing exotic derivatives whose payoffs are sensitive to tail behavior (e.g., deep out-of-the-money puts or calls), where a naive Gaussian assumption would systematically mis-price risk.

````python
import numpy as np
import plotly.graph_objects as go

# Solved MaxEnt parameters
lambda_1 = -0.1843
lambda_2 = 0.00114
lambda_3 = 0.4215
Z = 134.72
expected_value = 102.0

# Asset price range (avoid S_T=0 to prevent log/division errors)
S_T = np.linspace(0.1, 200, 1000)

# MaxEnt probability density function
# p*(S_T) = (1/Z) * S_T^(-lambda_3) * exp(-lambda_1*S_T - lambda_2*S_T^2)
pdf = (1.0 / Z) * (S_T ** (-lambda_3)) * np.exp(-lambda_1 * S_T - lambda_2 * (S_T**2))

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=S_T, y=pdf, mode='lines', name='MaxEnt Asset PDF',
        line=dict(color='#636EFA', width=3),
        hovertemplate='Asset Price: $%{x:.2f}<br>Probability Density: %{y:.5f}<extra></extra>'
    )
)

fig.add_trace(
    go.Scatter(
        x=[expected_value, expected_value], y=[0, max(pdf) * 1.05],
        mode='lines', name=f'Expected Value (${expected_value})',
        line=dict(color='#EF553B', width=2, dash='dash'),
        hovertemplate='Expected Forward Price: $102.00<extra></extra>'
    )
)

fig.update_layout(
    title=dict(
        text='<b>Maximum Entropy Risk-Neutral Asset Price Distribution</b><br><sup>Asymmetric crash risk fear modeled via log-space constraints</sup>',
        x=0.5, font=dict(size=16)
    ),
    xaxis=dict(title='Asset Price at Maturity $S_T$ ($)', gridcolor='rgba(200,200,200,0.2)', zeroline=False),
    yaxis=dict(title='Probability Density $p^*(S_T)$', gridcolor='rgba(200,200,200,0.2)', zeroline=False),
    template='plotly_white', width=900, height=550,
    legend=dict(yanchor="top", y=0.95, xanchor="right", x=0.95, bgcolor="rgba(255,255,255,0.6)")
)

fig.show()
```
````
